```python
import pandas as pd
import os
```

```python
file_path = r'D:\PLFS\National_Classification_of_Occupations2015.xlsx'

# Initialize an empty list to hold DataFrames
dfs = []

# Read the Excel file with all sheets
all_sheets = pd.read_excel(file_path, sheet_name=None)

# Process each sheet
for sheet_name, df in all_sheets.items():
    print(f"Sheet Name: {sheet_name}")
    try:
        df = pd.read_excel(file_path, sheet_name=sheet_name, usecols=[0, 1, 2, 3])
        
        # Drop rows where the first column is NaN
        df = df.dropna(subset=[df.columns[0]], how='all')
        df = df.dropna(subset=[df.columns[1]], how='all')
        df = df.dropna(subset=[df.columns[2]], how='all')
        
        # Append the processed DataFrame to the list
        dfs.append(df)
    except ValueError as e:
        print(f"Sheet Error: {sheet_name}")

# Concatenate all DataFrames in the list into one DataFrame
combined_df = pd.concat(dfs, ignore_index=True)
print(combined_df)
# # Define the output file path
# output_file_path = os.path.join(r'D:\PLFS\DivisionCode', 'output.xlsx')

# # Save the combined DataFrame to an Excel file
# combined_df.to_excel(output_file_path, index=False)

# print(f"Saved combined data to {output_file_path}")
```
